-- items adding to list
  - append(item)-it adds elements at end of list
  - extend(new_list)- it adds all new_list elements at end of the list
  - insert(pos,item)-it add the item at specified position

```python
l=[1,2,3]
l.append(105)
print(l)
l.extend([5,6,7])
print(l)
l.insert(0,100)
print(l)
```
