```python
# find the even number between 1-N and return it as List
n=int(input())
res=list()

for num in range(1,n+1,1):
    if num%2==0:
        res.append(num)
print(res)
```

```python



S = input()
S = S.lower()  # Handles uppercase letters

vowels = 0
consonants = 0

for char in S:
    if char.isalpha():  # Ignores spaces, numbers, and special characters
        if char in "aeiou":
            vowels += 1  # Adds to the count
        else:
            consonants += 1  # Fixed spelling and adds to the count

print("Vowels:", vowels)
print("Consonants:", consonants)
```

# WHILE LOOP

```python
# loop control statements

arr=[10,20,30,44,55,66]
target= int(input())
n=len(arr)
for i in range(n):
    if target== arr[i]:
        
        print(i)
        break
else:
    print(-1)
```

# Nested Loop:

```python
#Nested Loop Example

for i in range(1,5):
    for j in range(1,3):
        print(i,j)
```

```python
# Alternate rows
n=int(input())

for i in range(n):
    for j in range(n):
        if i%2==0:
            print("+", end= " ")
        else:
            print("-",end= " ")
    print()
```

```python
# LOW TRIANGLE
n=int(input())

for i in range(n):
    for j in range(n):
        if i >=j:
            
            print("*", end= " ")
        else:
            print(" ",end= " ")
    print()
```

```python
# Upper Triangle
n=int(input())

for i in range(n):
    for j in range(n):
        if i <=j:
            
            print("*", end= " ")
        else:
            print(" ",end= " ")
    print()
```

```python
# Upper Left Triangle
n=int(input())

for i in range(n):
    for j in range(n):
        if j<=n-i-1:
            print("*", end= " ")
        else:
            print(" ",end= " ")
    print()
```

```python
# Lower Right Triangle
n=int(input())

for i in range(n):
    for j in range(n):
        if j>=n-i-1:
            print("*", end= " ")
        else:
            print(" ",end= " ")
    print()
```

```python
# Lower Right Triangle
n=int(input())

for i in range(n):
    for j in range(n):
        if i==j:
            print("*", end= " ")
        else:
            print(" ",end= " ")
    print()
```

```python
# Secondary Diagonal

n = int(input())

for i in range(n):
    for j in range(n):
        if i + j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Lower Right Triangle
n=int(input())

for i in range(n):
    for j in range(n):
        if j<=n-i-1:
            
            print(n-j, end= " ")
        else:
            print(" ",end= " ")
    print()
```

```python
# HOLLOW SQUARE

n = int(input())

for i in range(n):
    for j in range(n):

        if i == 0 or i == n-1 or j == 0 or j == n-1:
            
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

```python
# HOLLOW SQUARE WITH DIAGONALS

n = int(input())

for i in range(n):
    for j in range(n):

        if i == 0 or i == n-1 or j == 0 or j == n-1 or i == j or i + j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```
