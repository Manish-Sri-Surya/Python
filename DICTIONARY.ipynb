```python
#get
details = {'name': 'manish','age':22,'location':'hyderabad'}
print(details.get('location'))
print(details.get('location','hyderabad'))
```

```python
details = {'name': 'manish','age':22,'course':'python'}

dict_keys = details.keys()
print(type(dict_keys))
print(dict_keys)
print(details.values())
print(details.items())
```

```python
#pop(), popitems()
details={'name':"mainsh",'age':22,"course":'python'}
print(details.pop("age"))
print(details)
print(details.popitem())
print(details)
```

```python
# union() 
s1={1,2,3,4,5}
s2={3,4,5,6,7}
print(s1.union(s2))
# intersection
print(s1.intersection(s2))
# intersection_update
s1.intersection_update(s2)
print(s1)
print(s2)
# difference
difference=s1.difference(s2)

print(difference)

print(s1.symmetric_difference(s2))
```

```python
s1={1,2,3,4,5}
s2={3,4,5,6,7}
# union
union= s1 | s2
print(union)

# intersection
intersection= s1 & s2
print(intersection)

#difference
difference= s1-s2
print(difference)

#symmertic_difference
symmetric_difference= s1 ^ s2
print(symmetric_difference)

```

```python
# issuesubset(), issuperset(), isdisjoint()
s1 = {1,2,3}
s2 = {1,2,3,4,5}
s3 = {7}

# is subset
print(s2.issubset(s1))
print(s1.issubset(s2))

print(set().issubset(s2))

# issuperset()
print(s2.issuperset(s2))
print(s1.issuperset(s2))

# isdisjointset()
print(s1.isdisjoint(s3))
```
