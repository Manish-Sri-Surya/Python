```python
# Input Formatting for All Data Types

# Integer Input
a = int(input("Enter integer value: "))

# Float Input
b = float(input("Enter float value: "))

# Complex Input
c = complex(input("Enter complex value: "))

# Boolean Input
d = bool(input("Enter boolean value: "))

# String Input
e = input("Enter string value: ")

# List Input
l = list(map(int, input("Enter list elements: ").split()))

# Tuple Input
t = tuple(map(int, input("Enter tuple elements: ").split()))

# Set Input
s = set(map(int, input("Enter set elements: ").split()))

# Dictionary Input
dic = {}

n = int(input("Enter number of items in dictionary: "))

for i in range(n):
    key = input("Enter key: ")
    value = int(input("Enter value: "))
    dic[key] = value

# Output

print("\nInteger =", a)
print("Float =", b)
print("Complex =", c)
print("Boolean =", d)
print("String =", e)
print("List =", l)
print("Tuple =", t)
print("Set =", s)
print("Dictionary =", dic)
```

```python
# output Formating

# 1, Using Old style Formating(using comma)
# name
# age
# My name is______ and i am a ______ years old
# Output Formatting

name = "manish"
age = 20

print("My name is " + name + " and I am " + str(age) + " years old")
```

```python
# % s(string)
# % d(int)
# % f(float)


name = "Manish"
age = 20
cgpa = 8.75

print("My name is %s, I am %d years old and my CGPA is %.2f" % (name, age, cgpa))
```

```python
# using Dot format

name = "Manish"
age = 20
cgpa = 8.75

print("My name is {} I am {} years old and my CGPA is {}" .format(name, age, cgpa))
```

```python
name = "Manish"
age = 20
cgpa = 8.75

print(f"My name is {name} I am {age} years old and my CGPA is:{cgpa:.2f}") 
```
