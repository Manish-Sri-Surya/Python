## STRING METHODS-
- case converstions methods
 - lower() [ converst uppercase to lower]
 - upper() [ convert upper to lower]
 - title() [convert first letter to upper]
 - captilize() [convert first leter to upper case an d remaing to lower]
 - swapcase() [ convert lower to upper & upper to lower]

```python
s="MAnISH Sri Surya"
print(s.lower())
print(s.upper())
print(s.title())
print(s.swapcase()) 
print(s.capitalize())
```

# White space or pattern removing
 .strip()- it removes the white space to both sides of the string
 
 .isstrip()-it removes the left side white space to the string
 
 .rstrip()-it removes the right side white space to the string

```python
s="        manish         "
print(s)
print(s.strip())
print(s.lstrip())
print(s.rstrip())
```

```python
s="####man#ish####"
print(s.strip("#"))
print(s.lstrip("#"))
print(s.rstrip("#"))
print(s)
```

```python
s="sscacccacbsacCd"

print(s.count("s"))
print(s.count("cac"))
# count the all 'c' (upper+lower)

print(s.count("c")+s.count("C"))
# convert lower and find count of "c"
str1=s.lower()
print(str1.count("c"))
```

# string checking methods

- islower()-it returns True if entire string is in lower case
- isupper()-it returns true if entire string is in upper case
- istitle()-it returns True if entire  string is in title format
- iscaptilize()- it returns True if entire string is in captilize format 
- isdigit()- it returns True if entire string contains all digits
- isaplha()- it returns True if entire string contains all alphabets
- isalnum()- it returns True if entire string contains either digits or alphabets
- startswith(sub_string) - it returns True if  string start with substring
- endswith(sub_string) - it returns True if  string ends with digits



```python
print("code".islower())

print("CODE".isupper())

print("1234".isdigit())

print("Manish".istitle())

print("123a".isalnum())

print("abc123".isalpha())

print("Codegnan".startswith("Co"))

```
